# FleetSplit figures

Thin wrapper over `analysis/figures.py` - figures are scripted, never hand-made, and every
metric is a query over the SQLite run DBs. Run the measurement sequence first
(`make preflight` on the dev machine, or README.md on hardware).

Interactive use: `uv sync --extra analysis`, then pick this venv as the kernel.

```python
import json, sys, pathlib
ROOT = pathlib.Path.cwd().parent if pathlib.Path.cwd().name == 'analysis' else pathlib.Path.cwd()
sys.path.insert(0, str(ROOT))
from analysis.figures import OUT, POLICY_COLORS, newest
OUT.mkdir(parents=True, exist_ok=True)
runs = ROOT / 'runs'
sweep = newest(runs, '*_sweep.db')
policy_dbs = {k: db for k in POLICY_COLORS if (db := newest(runs, f'*_{k}.db'))}
sweep, policy_dbs
```

```python
from analysis.figures import latency_by_cut, stage_breakdown, policy_cdfs, tables
cuts_path = ROOT / 'model' / 'artifacts' / 'cuts.json'
latency_by_cut(sweep)
stage_breakdown(sweep)
if policy_dbs:
    policy_cdfs(policy_dbs)
tables(sweep, policy_dbs, json.loads(cuts_path.read_text()) if cuts_path.exists() else {})
```

```python
from IPython.display import Image, display
for name in ('latency_by_cut', 'stage_breakdown', 'policy_cdfs'):
    if (OUT / f'{name}.png').exists():
        display(Image(str(OUT / f'{name}.png')))
```
